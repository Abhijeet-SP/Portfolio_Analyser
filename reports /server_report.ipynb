Many operation are there for with open("file_name", "operation_type") as file:

1. "r" read only operation. [file.read() operation]
2. "w" write only operation (can overwrite). [file.write() operation]
3. "a" append only operation. [file.write() operation for append]
4. "w+" file is ready to write and read. if file doesn't exist it will be created, if existed then it will be truncated. 

Concpet is a "cursor" points to the last read and write line. So to read again we need [file.seek(0) - set cursor to the first line to read again]
